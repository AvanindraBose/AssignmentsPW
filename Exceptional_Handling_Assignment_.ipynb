```python
# Question 1
# Multithreading vs. Multiprocessing
# Multithreading:

# Preferred for I/O-bound tasks (e.g., file I/O, database queries, network requests).
# Threads share the same memory space, avoiding overhead related to creating new processes.
# Allows better performance where the Global Interpreter Lock (GIL) does not limit concurrency (e.g., waiting for I/O).
# Multiprocessing:

# Preferred for CPU-bound tasks (e.g., numerical computations, data processing).
# Each process has its own memory space, bypassing the GIL and allowing true parallelism.
# Overheads of process creation and inter-process communication (IPC) are acceptable in exchange for better parallelism.
```

```python
#  Question2
# A process pool is a collection of pre-initialized worker processes managed by a pool manager. It is used to execute tasks in parallel efficiently.

# Benefits:
# Reduces the overhead of creating/destroying processes for every task.
# Simplifies managing a fixed number of processes.
# Provides methods like map, apply, and starmap for distributing tasks among workers.
```

```python
#  Question3
# Multiprocessing is a module in Python that supports the creation of processes, allowing programs to achieve parallelism.

# Uses:
# Bypass GIL for CPU-bound tasks.
# Divide work across multiple CPU cores for faster execution.
```

```python
#  Question 4
import threading
import time

numbers = []
lock = threading.Lock()

def add_numbers():
    for i in range(10):
        with lock:
            numbers.append(i)
            print(f"Added: {i}, List: {numbers}")
        time.sleep(0.1)

def remove_numbers():
    for _ in range(10):
        with lock:
            if numbers:
                removed = numbers.pop(0)
                print(f"Removed: {removed}, List: {numbers}")
        time.sleep(0.15)

thread1 = threading.Thread(target=add_numbers)
thread2 = threading.Thread(target=remove_numbers)
thread1.start()
thread2.start()
thread1.join()
thread2.join()

```

```python
#  Question 5
# Threads:

# threading.Lock: Prevents race conditions by ensuring only one thread accesses a shared resource at a time.
# queue.Queue: A thread-safe data structure for passing data between threads.
# Processes:

# multiprocessing.Queue: Shares data between processes safely.
# multiprocessing.Manager: Creates shared data structures like lists and dictionaries.
# multiprocessing.Value and multiprocessing.Array: Share individual values or arrays between processes.
```

```python
#  Question 6
# Why Crucial:

# Unhandled exceptions can lead to data corruption, deadlocks, or terminated threads/processes.
# Debugging concurrent programs is harder without proper error handling.
# Techniques:

# Try-Except Blocks: Wrap thread/process logic in try-except.
# Thread-Specific Logging: Log errors with thread/process IDs.
# concurrent.futures: Handles exceptions and retrieves them in the main thread via result().
```

```python
#  Question 7
from concurrent.futures import ThreadPoolExecutor
import math

def factorial(n):
    return f"Factorial of {n} is {math.factorial(n)}"
with ThreadPoolExecutor() as executor:
    results = executor.map(factorial, range(1, 11))

for result in results:
    print(result)
```

```python
# Question 8
from multiprocessing import Pool
import time

def square(n):
    return n ** 2
for pool_size in [2, 4, 8]:
    start_time = time.time()
    with Pool(pool_size) as pool:
        results = pool.map(square, range(1, 11))
    end_time = time.time()

    print(f"Pool Size: {pool_size}, Squares: {results}, Time: {end_time - start_time:.4f} seconds")
```
